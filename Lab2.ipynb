## **1. Human Visual Perception: Mach Bands**
**Exercise 1.1:** Simulating Mach bands to show how our eyes perceive transitions.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create Mach Bands
bands = np.zeros((100, 600))
for i in range(6):
    bands[:, i*100 : (i+1)*100] = i*40

plt.imshow(bands, cmap='gray')
plt.title("Synthetic Mach Bands")
plt.axis('off')
plt.show()
```

## **2. Simultaneous Contrast**
**Exercise 2.1:** Perceived brightness depends on the background.

```python
img = np.ones((200, 600)) * 100
img[:, :200] = 20
img[:, 400:] = 200
img[50:150, 250:350] = 120

plt.imshow(img, cmap='gray')
plt.title("Simultaneous Contrast Example")
plt.axis('off')
plt.show()
```

## **3. Sampling (Spatial Resolution)**
**Exercise 4.1:** Reducing the number of pixels in an image.

```python
original = image.copy()
down_64 = original[::4, ::4]
down_32 = original[::8, ::8]

plt.figure(figsize=(9,3))
plt.subplot(1,3,1)
plt.imshow(original, cmap='gray')
plt.title("256x256")
plt.subplot(1,3,2)
plt.imshow(down_64, cmap='gray')
plt.title("64x64")

plt.subplot(1,3,3)
plt.imshow(down_32, cmap='gray')
plt.title("32x32")

plt.show()
```

## **4. Quantisation (Intensity Resolution)**
**Exercise 5.1:** Reducing the number of grey levels.

```python
def quantise(img, levels):
    step = 256 // levels
    return (img // step) * step

q256 = quantise(image, 256)
q16 = quantise(image, 16)
q4 = quantise(image, 4)

plt.figure(figsize=(9,3))
plt.subplot(1,3,1)
plt.imshow(q256, cmap='gray')
plt.title("256 levels")

plt.subplot(1,3,2)
plt.imshow(q16, cmap='gray')
plt.title("16 levels")

plt.subplot(1,3,3)
plt.imshow(q4, cmap='gray')
plt.title("4 levels")

plt.show()
```

## **6. Distance Measures (Exercise 8)**
**Task:** Implement the formulas for Euclidean, City-block ($D_4$), and Chessboard ($D_8$) distances.

```python
import math

p = (2, 3)
q = (7, 9)


D_euclid = math.sqrt((p[0] - q[0])**2 + (p[1] - q[1])**2)

D4 = abs(p[0] - q[0]) + abs(p[1] - q[1])

D8 = max(abs(p[0] - q[0]), abs(p[1] - q[1]))

print(f"Euclidean: {D_euclid:.2f}, D4: {D4}, D8: {D8}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# start with random grayscale image
image = np.random.randint(0, 256, (256, 256))

# Spatial Resolution(Sampling)
sample_128 = image[::2, ::2]
sample_64 = image[::4, ::4]
sample_32 = image[::8, ::8]

plt.figure()
plt.imshow(sample_128, cmap='gray')
plt.title("128x128")
plt.show()

plt.figure()
plt.imshow(sample_64, cmap='gray')
plt.title("64x64")
plt.show()

plt.figure()
plt.imshow(sample_32, cmap='gray')
plt.title("32x32")
plt.show()


# Intensity Resolution(Quantisation)
def quantise(img, levels):
    step = 256 // levels
    return (img // step) * step

q256 = quantise(image, 256)
q16 = quantise(image, 16)
q4 = quantise(image, 4)

plt.figure()
plt.imshow(q16, cmap='gray')
plt.title("16 levels")
plt.show()

plt.figure()
plt.imshow(q4, cmap='gray')
plt.title("4 levels")
plt.show()
```

Reflection questions

1. Why is a digital image always an approximation of reality?

because real scenes are continuous, but digital images are made of pixels and limited intensity values, When we sample and quantise, we lose some details
So the image is never exactly the same as reality.


2. How does human perception complicate objective image analysis?

Human perception depends on contrast and surroundings, not just pixel values
Sometimes we see differences that are not actually in the data, like in Mach bands So what we see is not always equal to the actual numbers.


3. Why is interpolation unavoidable in geometric transformations?

When we scale or rotate an image, the new pixel positions do not match the original grid, So we need interpolation to estimate the missing values ,Without it the image would look broken or have gaps.
